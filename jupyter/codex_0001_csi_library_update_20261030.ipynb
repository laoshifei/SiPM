3. **再比较信号期响应候选。** 对有限个预先列明的读出传递函数变化、低频形状波动与短尺度残差候选，联合检查均ADC、三种PID的中心和完整协方差、F50/102/202及额外尾部。分量比例波动与读出极点变化仍是竞争解释。校准选择和不确定度应在各训练块内传播，模型冻结后再用独立Geant4母事件检查模拟统计；增加重放不能增加独立输运历史。
